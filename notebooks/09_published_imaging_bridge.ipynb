**Accuracy and agreement**

*Notation.* $P$ is the set of voxels labelled by the prediction (here the SegResNet), $R$ the set
labelled by the reference mask, and $\Omega$ all brain voxels in the image; $\lvert\cdot\rvert$ counts
voxels and $v$ is the volume of one voxel in mm³. The four voxel counts of the confusion matrix are

$$
\mathrm{TP}=\lvert P\cap R\rvert,\qquad
\mathrm{FP}=\lvert P\setminus R\rvert,\qquad
\mathrm{FN}=\lvert R\setminus P\rvert,\qquad
\mathrm{TN}=\lvert\Omega\setminus(P\cup R)\rvert ,
$$

true positives, false positives, false negatives and true negatives. Each metric below is computed
separately for each compartment (WT, TC, ET) and each subject.

| Term | Full name | Meaning | Range and units |
| --- | --- | --- | --- |
| Dice | Dice similarity coefficient (Sørensen–Dice) | Volume overlap of $P$ and $R$, relative to their mean size; insensitive to where an error lies. | 0 (disjoint) to 1 (identical); unitless |
| HD95 | 95th-percentile Hausdorff distance | Boundary distance: the distance within which 95% of the surface points of each mask lie from the other mask's surface. The ordinary Hausdorff distance uses the maximum (100%) and is dominated by a single outlier. | 0 upwards; mm |
| VE | Absolute volumetric error | Difference in total volume, regardless of where the voxels are: two masks can have VE = 0 and no overlap. | 0 upwards; mm³ |
| Sensitivity | Sensitivity, true-positive rate, recall | Share of reference voxels that the prediction also labels; low values mean under-segmentation. | 0 to 1 |
| Specificity | Specificity, true-negative rate | Share of non-reference voxels that the prediction leaves unlabelled; with about $10^6$ background voxels per subject it saturates near 1 and does not discriminate. | 0 to 1 |
| CCC | Lin's concordance correlation coefficient ([Lin 1989](https://doi.org/10.2307/2532051)) | Agreement of paired values with the identity line $y = x$: penalizes both scatter (as the Pearson correlation does) and systematic offset or scale differences (which the Pearson correlation ignores). | −1 to 1; 1 = perfect agreement |
| Bland–Altman limits | 95% limits of agreement ([Bland & Altman 1986](https://doi.org/10.1016/S0140-6736%2886%2990837-8)) | The interval expected to contain about 95% of the differences between the two methods, assuming roughly normal differences. | Units of the measurement (here pp) |
| pp | Percentage points | Arithmetic difference between two percentages: 12% − 10% = 2 pp, not a 20% relative change. | pp |
| Hit-plot cell value | Percentage of region occupied | For subject $s$, region $r$ (with voxel set $A_r$) and compartment $C$: the share of the region covered by that compartment. | 0 to 100%; the unit of Table 2 |
| Metrics Reloaded | — | A consensus framework for choosing validation metrics according to the problem ([Maier-Hein et al. 2024](https://doi.org/10.1038/s41592-023-02151-z)). | — |

*Definitions.* Overlap, volume and voxel-classification metrics:

$$
\mathrm{Dice}=\frac{2\,\lvert P\cap R\rvert}{\lvert P\rvert+\lvert R\rvert}
=\frac{2\,\mathrm{TP}}{2\,\mathrm{TP}+\mathrm{FP}+\mathrm{FN}},\qquad
\mathrm{VE}=v\,\bigl\lvert\,\lvert P\rvert-\lvert R\rvert\,\bigr\rvert ,
$$

$$
\mathrm{Sensitivity}=\frac{\mathrm{TP}}{\mathrm{TP}+\mathrm{FN}}=\frac{\lvert P\cap R\rvert}{\lvert R\rvert},
\qquad
\mathrm{Specificity}=\frac{\mathrm{TN}}{\mathrm{TN}+\mathrm{FP}} .
$$

Boundary distance. Let $\partial P$ and $\partial R$ be the surface points of the two masks (in
millimetre coordinates) and $d(x,S)=\min_{y\in S}\lVert x-y\rVert_2$ the distance from a point $x$ to
the nearest point of a set $S$. With $Q_{95}$ the 95th percentile over the points named beneath it,

$$
\mathrm{HD95}(P,R)=\max\Bigl\{\;\underset{x\in\partial P}{Q_{95}}\;d(x,\partial R),\;\;
\underset{y\in\partial R}{Q_{95}}\;d(y,\partial P)\Bigr\},
\qquad
\mathrm{HD}(P,R)=\max\Bigl\{\max_{x\in\partial P}d(x,\partial R),\;\max_{y\in\partial R}d(y,\partial P)\Bigr\}.
$$

This is the convention of the synthetic examples in this notebook (the larger of the two directed
95th percentiles, as in MONAI); tools differ in how surfaces are extracted and how the two directions
are pooled, which is why the published values are not recomputed here.

Agreement between paired values. For $n$ pairs $(x_i, y_i)$ (here hit-plot cell values from the
SegResNet and the reference masks; $n = 8600$ cells per compartment), with means $\bar x,\bar y$,
variances $s_x^2, s_y^2$ and covariance $s_{xy}$, and differences $d_i = x_i - y_i$ with mean
$\bar d$ and standard deviation $s_d$:

$$
\rho_c=\frac{2\,s_{xy}}{s_x^2+s_y^2+(\bar x-\bar y)^2},
\qquad
\text{limits of agreement}=\bar d\pm 1.96\,s_d .
$$

The hit-plot cell value itself is

$$
h_{s,r,C}=100\cdot\frac{\lvert C_s\cap A_{s,r}\rvert}{\lvert A_{s,r}\rvert}\ \ [\%],
$$

where $C_s$ is the compartment's voxel set and $A_{s,r}$ the voxel set of region $r$, both in subject
$s$'s native space. A difference between two such values is therefore in percentage points of the
region's volume.

## Background: from a tumour mask to a patient's cognition

**Neuroscience and clinical track.**
- *Why segment at all.* Volumes and their change over time underlie RANO response assessment,
  surgical extent-of-resection reports and radiotherapy target volumes. Manual segmentation is slow
  and variable, so a validated automatic method makes regional, longitudinal description feasible.
- *What the study added.* It combined a public deep-learning segmenter with subject-specific
  FreeSurfer parcellation in each patient's own anatomy, giving for every patient a table of which
  regions each tumour compartment touches ("hit-plots"), and reported accuracy against reference
  masks on a leak-safe cohort ([Lundervold et al. 2026](https://doi.org/10.1080/03091902.2026.2713703)).
- *Why location is not enough.* A tumour in the left supramarginal gyrus does not simply remove
  "the supramarginal function". Cognition depends on networks of regions linked by white matter, and
  glioma cells infiltrate along that white matter well beyond the enhancing rim. Recent cohorts find
  that, outside a few critical regions, high-grade gliomas "impair cognition generally"
  ([Houskamp et al. 2025](https://doi.org/10.3390/cancers17172775)).
- *Why it matters at the bedside.* Surgeons balance extent of resection against function; radiation
  oncologists add margins for invisible tumour; neuropsychologists measure what the patient can do.
  Each needs a different description of the tumour–brain interface.

**Modeling track.**
- *Segmentation as estimation.* A segmenter estimates a label field from images; Dice and HD95
  summarize its disagreement with another estimate (the reference), not with the truth.
- *Hit-plots as a projection.* Intersecting labels with a parcellation projects the tumour onto a
  fixed set of regions: a vector of overlaps per subject and compartment. Like any projection, it
  depends on boundaries, both the tumour's and the regions'.
- *Visible versus true extent.* Imaging sees densities above a threshold. Reaction–diffusion models
  connect the visible boundary to an unobserved density tail ([Swanson et al. 2008](https://doi.org/10.1038/sj.bjc.6604125)); the
  article itself names growth models as a natural extension.
- *From regions to networks.* Graph models of structural and functional connectivity
  ([Honey et al. 2009](https://doi.org/10.1073/pnas.0811168106)) and disconnection indices ([Herbet et al. 2024](https://doi.org/10.1093/brain/awae093)) turn a
  regional vector into a network perturbation, which is what cognitive models need as input.

**How to read the figures.** Every figure caption begins with its evidence status (published or
synthetic). Figures 1–3 are constructed: regions A, B, C and S are not anatomical, and the growth
cases are prescribed. Read each with three questions: what is measured, in which units, and what
would change the conclusion.

## 1. The published study: motivation, methods and results

**Motivation.** The article asks whether a public, reproducible chain of open tools can segment a
glioma and its surroundings precisely enough to describe, in each patient's own anatomy, which regions
the tumour involves. It distinguishes three senses of "precision" (section 2.1 of the article):
1. *segmentation accuracy*: agreement of compartment masks with a reference (Dice, HD95 and others);
2. *anatomical localisation*: naming a region set or laterality;
3. *anatomical profiling*: the per-subject overlap between compartments and anatomical regions
   (hit-plots).

**Methods.**
- *Images:* co-registered T1, T1c, T2 and FLAIR.
- *Segmentation:* the MONAI Bundle 0.4.8 BraTS three-compartment SegResNet, a trained deep-learning
  model, behind a thin adapter that maps {T1, T1c, T2, FLAIR} to {NCR, ED, ET}.
- *Anatomy:* FreeSurfer 8.2.0 recon-all-clinical, which uses SynthSeg and SynthSR, providing the
  wmparc parcellation of 172 regions in each subject's native space, without warping to an atlas.
- *Hit-plots:* overlap of each compartment with each region; volumetric, probabilistic and
  functional-anatomy variants.
- *Cohort:* 50 IDH-wildtype WHO grade 4 glioblastomas from UCSF-PDGM, chosen to exclude subjects in
  the BraTS 2021 training data (leak-safe), stratified by sex and survival tertile; 20 women and
  30 men, median age 64 years (range 29–83). One public LUMIERE patient with six visits illustrates a
  longitudinal course.

The reference is the dataset mask, not histological truth. This notebook performs no new leakage
audit.

**Reading Table 1.** Each cell gives the median over 50 subjects and, in brackets, the interquartile
range (Q1 to Q3). The abstract rounds the medians to Dice 0.90, 0.94 and 0.86 and HD95 4.1, 2.2 and
2.0 mm. All three median Dice values lie above the BraTS 2018 "strict" band the article uses
(WT ≥ 0.85, TC ≥ 0.75, ET ≥ 0.70), yet 16 of 50 subjects were flagged against that band in a
pre-registered audit and all 16 were retained after visual review (section 3.1). The median
hides the spread: a quarter of subjects have a WT HD95 above 6.87 mm.

**Neuroscience and clinical track.**
- *What a median WT HD95 of 4.12 mm means.* For half the patients, 95% of the predicted tumour
  surface lies within about 4 mm of the reference surface. That is comparable to a typical PTV margin
  (≤ 3 mm with image guidance) but much smaller than the 15 mm CTV margin
  ([Niyazi et al. 2023](https://doi.org/10.1016/j.radonc.2023.109663)).
- *Volume error.* A median absolute WT volume error of 8170 mm³ (8.2 mL) is large in absolute terms
  but small relative to typical glioblastoma volumes of tens of millilitres; it matters most when
  following small residual tumours.
- *Sensitivity is lowest for ET (0.793).* Thin enhancing rims are where small boundary disagreements
  cost the most.

**Modeling track.**
- *Why report IQRs.* Segmentation metrics are skewed and bounded; a median with quartiles describes
  the distribution better than a mean with a standard deviation.
- *Why specificity saturates.* With about 10⁶ non-tumour voxels and about 10³ false positives per
  subject, specificity rounds to 1.000 even for clear over-segmentation, as the article notes. It is
  not informative here; Metrics Reloaded makes the same point about class imbalance
  ([Maier-Hein et al. 2024](https://doi.org/10.1038/s41592-023-02151-z)).
- *Reference noise.* Disagreement with a reference contains both model error and reference error;
  the article discusses one subject (sub-0396) as a worst-case boundary disagreement between
  segmenters.

## 2. What Dice and HD95 measure

Let $P$ be the predicted mask and $R$ the reference mask.

**Dice** measures overlap:
$$\text{Dice}=\frac{2\,|P\cap R|}{|P|+|R|},$$
from 0 (no overlap) to 1 (identical). It counts voxels, so an error matters in proportion to its
volume, wherever it is.

**HD95** measures boundary distance. For every surface point of $P$, find the distance to the nearest
surface point of $R$, and vice versa. The Hausdorff distance is the largest of these distances; HD95 is
the 95th percentile instead, which ignores the worst 5% of surface points. It is in millimetres and
depends on *where* an error is, not on its volume.

The two are complementary. A small error far away barely changes Dice but can dominate HD95 once it
contributes more than 5% of the surface points. A thin rim of error all around the boundary lowers
Dice noticeably while HD95 stays at the rim's thickness. In Table 1, the median WT Dice of 0.901 and
HD95 of 4.12 mm therefore describe different aspects of the same segmentations.

**Neuroscience and clinical track.** Dice answers "how much of the tumour volume agrees?"; HD95
answers "how far off is the boundary, at worst, for most of it?". A surgeon planning a margin or a
radiation oncologist drawing a CTV cares about the boundary; a volumetric response assessment cares
about volume.

**Modeling track.** Dice is a set-overlap ratio insensitive to error location; HD95 is a robust
quantile of a distance distribution and insensitive to error volume. Neither is differentiable in
its usual form; training losses use soft Dice, and boundary losses approximate distance terms.

**Convention used here.** Surface = mask voxels with a face neighbour outside the mask (in-plane only
for a single slice). HD95 = the larger of the two directed 95th percentiles. Other tools differ in the
surface definition and in how the two directions are pooled, so the synthetic numbers below are not
comparable to the published table.

## 6. The tumour–brain interface: the visible boundary is not the edge

**Why this section.** Sections 2–5 treated the reference mask as the tumour. But a diffuse glioma
has no sharp edge. MRI shows only where cell density is high enough to change the signal: enhancement
where the blood–brain barrier is disrupted, FLAIR abnormality where oedema and infiltration raise
water content. Beyond both, tumour cells continue at lower density. A hit-plot therefore depends on
*which* boundary is used, and the same visible tumour can hide very different amounts of invisible
tumour.

**Evidence that the interface is graded.**
- *Diffusion imaging* shows a gradual change through the peritumoural zone of glioblastoma,
  compatible with infiltration accentuated near the tumour, in contrast to the predominantly vasogenic
  oedema around metastases (30 glioblastomas, 28 metastases;
  [Würtemberger et al. 2024](https://doi.org/10.1007/s00062-024-01416-0)).
- *Radiomic infiltration maps* from standard MRI in UPenn-GBM and UCSF-PDGM (the cohort source of
  the published study; 872 subjects) found that high-infiltration peritumoural regions had lower
  fractional anisotropy and higher perfusion, and that infiltration burden and radial extent were
  associated with survival ([Cepeda et al. 2026](https://doi.org/10.1016/j.ejrad.2026.112888)).
- *Recurrence patterns.* In 390 paired MRIs, a deep-learning study from our group found that a low
  ratio of enhancing core to non-enhancing tumour volume (≤ 0.324) was associated with longer overall
  survival (adjusted HR 0.56), that longer time to progression went with larger HD95 distances between
  primary and recurrent tumour, and that occipital tumours most often migrated to new sites (57.1%);
  the authors suggest these features may inform personalized radiotherapy margins
  ([Hannisdal et al. 2026](https://doi.org/10.1038/s41698-026-01557-0)).

**Clinical practice.** For glioblastoma, the ESTRO-EANO guideline defines the CTV as the resection
cavity and residual enhancing tumour plus a 15 mm margin, reduced from 20 mm, adjusted to anatomical
barriers ([Niyazi et al. 2023](https://doi.org/10.1016/j.radonc.2023.109663)). In surgery, the RANO resect group found that resection
beyond the enhancing tumour into the FLAIR abnormality (supramaximal resection) was associated with
longer survival only in patients younger than 65 years with newly diagnosed disease (40 versus
20 months), in a retrospective series of 1260 patients ([Teske et al. 2026](https://doi.org/10.1093/neuonc/noaf239)). A smaller
retrospective study classified tumours by Swanson's ρ/D ratio and reported that moderately diffuse,
MGMT-methylated tumours benefited from supramaximal resection ([Tropeano et al. 2025](https://doi.org/10.3390/cancers17060995)).
Both are observational: they show associations, not that the extra resection caused the benefit.

**A minimal model.** The proliferation–invasion (Fisher–Kolmogorov) equation describes the
normalized tumour-cell density $c(x,t)$ along a line from the tumour centre:

$$\frac{\partial c}{\partial t}=D\,\frac{\partial^2 c}{\partial x^2}+\rho\,c\,(1-c).$$

Diffusion with coefficient $D$ spreads cells; logistic proliferation at rate $\rho$ fills space up to
the maximum density 1. A travelling front forms with width about $\sqrt{D/\rho}$ and speed about
$2\sqrt{D\rho}$. Following [Swanson et al. 2008](https://doi.org/10.1038/sj.bjc.6604125), we read T1Gd as abnormal where
$c \ge 0.80$ and T2/FLAIR where $c \ge 0.16$, and add a 1% level to show the low-density tail.

**What the next cell does.**
1. It grows three tumours with the same proliferation rate, ρ = 15/year, and three diffusivities from
   the range measured in 32 untreated glioblastomas by [Wang et al. 2009](https://doi.org/10.1158/0008-5472.CAN-08-3863): 3.2 (lowest),
   29.7 (median) and 468.9 mm²/year (highest). Pairing the extremes of D with the median ρ is our
   construction.
2. Each starts as a small seed (density 0.5 within 1 mm of the centre) and is stopped when its T1Gd
   boundary reaches 15 mm, so all three look the same on contrast-enhanced MRI.
3. It reports where each visibility level is crossed, the share of tumour cells beyond the FLAIR
   boundary and beyond a 15 mm margin around the T1Gd boundary (a CTV-like margin, not a treatment
   plan), and which of the constructed regions A (0–25 mm), B (25–35 mm) and C (35–120 mm) each
   level reaches.

The model is planar (a line, not a sphere), isotropic and homogeneous: no white-matter guidance, no
mass effect, no treatment. The explicit time step respects the stability limit
$\Delta t \le \Delta x^2/(2D)$.

  distance from T1Gd to FLAIR boundary is about six front widths (5.9, 5.5 and 6.1), so the gap
  between the two visible boundaries carries information about $\sqrt{D/\rho}$; this is the idea
  behind Swanson's estimates of D and ρ from two contrasts.
- *Invisible tail.* The 1% level lies 4.5, 12.9 and 53.7 mm beyond the T1Gd boundary.
- *Beyond a 15 mm margin.* No cells for the low-diffusivity tumour, 0.02% for the median and 25.6%

- *Hit-maps depend on the level.* At the FLAIR level the high-diffusivity tumour already reaches

- *Starting assumption.* The high-diffusivity front has not settled into its final shape when its
  T1Gd boundary reaches 15 mm (after about 143 days of model time; its centre has reached only 0.87 of
  the maximum density), so its numbers depend on the seed:
  a 3 mm seed instead of 1 mm moves its FLAIR boundary inward by 3.4 mm and lowers the share beyond the
  margin from 25.6% to about 21.6%. The other two cases change by 0.2 mm or less. Halving the grid
  spacing changes no boundary by more than 0.25 mm.

**Neuroscience and clinical track.**
1. *Why one margin may not fit all.* With identical enhancing tumours, a fixed 15 mm margin covers
   all modelled cells of a nodular tumour but leaves a quarter of the cells of a very diffuse one
   outside. This is the reasoning behind proposals for individualized margins and supramaximal
   resection in selected patients; the studies above show associations, not proof of benefit.
2. *Why FLAIR is not "just oedema".* In glioblastoma, the FLAIR abnormality contains infiltrating
   cells; the hit-plot of published Figure 4c is dominated by this compartment in white matter.
3. *Why recurrences appear at the margin.* Cells below the visibility levels are neither seen nor
   resected; the model makes that intuition quantitative, but only for its stated assumptions.

**Modeling track.**
1. *Identifiability.* A single T1Gd boundary cannot distinguish the three cases; two boundaries (or
   two time points) constrain $\sqrt{D/\rho}$ and $\sqrt{D\rho}$ respectively. Notebook 03 (section 9)
   fits D and ρ to one interval of real LUMIERE data and shows what that interval can and cannot
   identify.
2. *Planar versus spherical.* In three dimensions an outer shell holds more volume per millimetre than
   an inner one, so for the same density profile the share of cells beyond a margin would be larger.
3. *Anisotropy.* Real infiltration follows white matter; diffusion-tensor-guided models replace the
   scalar D by a tensor. The fraction beyond a margin then depends on direction, which is exactly what
   a hit-plot of white-matter regions begins to describe.
4. *Thresholds are assumptions.* The levels 0.80 and 0.16 are a modelling convention, not measured
   detection limits; changing them shifts every boundary in Table 5.

## 7. From local to networks: why a regional map is not a cognitive map

**Why this section.** A hit-plot answers "where is the tumour?". Clinicians and patients ask "what
will it do to me?". Between the two lies how the brain computes, which is distributed across
networks rather than housed in single regions.

**Connectionism, briefly.** [McCulloch & Pitts (1943)](https://doi.org/10.1007/BF02478259) modelled neurons as threshold
units whose networks can compute logical functions. Learning by adjusting connection weights through
back-propagated errors ([Rumelhart, Hinton & Williams 1986](https://doi.org/10.1038/323533a0)) became the basis of deep
learning ([LeCun, Bengio & Hinton 2015](https://doi.org/10.1038/nature14539)). The SegResNet that produced the masks in
section 1 is a direct descendant: millions of weighted connections between simple units, trained
end to end. In the connectionist view of the brain, cognition arises from patterns of action potentials
and their integration at synapses across large networks. That view explains a great deal; sections 7–8
also note what it leaves out.

**Structural and functional connectivity.** In vivo, *structural connectivity* (SC) is estimated from
diffusion MRI tractography and *functional connectivity* (FC) from correlated BOLD fluctuations.
Structure constrains function: SC predicts part of resting-state FC, including FC between regions
without a direct structural connection ([Honey et al. 2009](https://doi.org/10.1073/pnas.0811168106)). Notebooks 11 and 12 build
both kinds of network, on a phantom and on one patient.

**Disconnection.** Damage to white matter can impair a function by cutting communication between
intact regions, an idea with a long history ([Catani & ffytche 2005](https://doi.org/10.1093/brain/awh622)). For glioma surgery,
[Herbet, Duffau & Mandonnet (2024)](https://doi.org/10.1093/brain/awae093) propose *connectotomy*: describe a resection by a
handful of indices of how much each network's structural connectivity is cut, rather than by
thousands of voxels, and add the patient's structure–function phenotype and plasticity as predictors.

**Glioma evidence, 2023–2026.**
- *Cognition is affected generally.* In 44 patients with glioblastoma, larger tumours predicted worse
  attention, memory, language and visuospatial scores and left-hemisphere involvement predicted worse
  attention, language and memory; lesion–symptom mapping found only one specific region (around the
  left sagittal stratum, for language) ([Houskamp et al. 2025](https://doi.org/10.3390/cancers17172775)).
- *One-to-many mappings.* In 252 patients operated awake for low-grade glioma, multilayer network
  analysis linked tasks to combinations of cortical damage and disconnection; tasks such as verbal
  fluency were better explained by several disconnection mechanisms than by one region
  ([Nave et al. 2026](https://doi.org/10.1016/j.nicl.2026.104052)).
- *Classical maps revised.* In 115 patients with left-hemisphere gliomas, the anterior temporal lobe
  affected both comprehension and production, challenging the Broca-production/Wernicke-comprehension
  dichotomy; the authors caution that intraoperative single-word naming may miss regions critical for
  sentences ([Ntemou et al. 2026](https://doi.org/10.1093/cercor/bhag112)).
- *Tumours join networks.* In intracranial recordings during language tasks, task-relevant responses
  activated tumour-infiltrated cortex well beyond the regions normally recruited, and the degree of
  functional connectivity between glioblastoma and the rest of the brain was negatively associated with
  both survival and language performance ([Krishna et al. 2023](https://doi.org/10.1038/s41586-023-06036-1)). Glioma-infiltrated
  cortex still synchronizes during speech but recruits a diffuse network and carries less information
  ([Aabedi et al. 2021](https://doi.org/10.1073/pnas.2108959118)).
- *Effects beyond the tumour.* In 31 IDH-wildtype glioblastomas, contralesional cortical surface area
  and connectivity metrics were associated with survival and Ki-67
  ([Styliara et al. 2025](https://doi.org/10.1007/s00234-025-03683-1)).
- *Syntheses.* Reviews argue that glioma is better understood with networks than with
  localizationist thinking: symptom networks, the connectome and tumour-cell networks
  ([Douw, Reijneveld & Mandal 2026](https://doi.org/10.1038/s41582-025-01171-x)); and that connectome-informed surgery, prognosis and
  rehabilitation "remain investigational" ([Vargas et al. 2026](https://doi.org/10.1016/j.clineuro.2026.109475)).

**Critical voices.**
- *Lesion network mapping.* [van den Heuvel et al. (2026)](https://doi.org/10.1038/s41593-025-02196-7) re-examined LNM studies
  and argued that LNM repeatedly samples one normative FC matrix, so that lesions, synthetic or random
  maps converge on similar, nonspecific networks. LNM's developers replied, in a preprint not yet peer
  reviewed, that specificity testing across 1090 lesion locations from 34 studies confirms
  lesion–deficit specificity ([Siddiqi et al. 2026](https://doi.org/10.64898/2026.02.24.707529)). Commentaries followed on null
  models ([Zalesky & Cash 2026](https://doi.org/10.1038/s41593-026-02259-3)) and on biological limitations
  ([Pini, Salvalaggio & Corbetta 2026](https://doi.org/10.1038/s41593-026-02319-8)). The debate is open.
- *BOLD near tumours.* Task and resting-state BOLD can localize eloquent cortex, but neurovascular
  uncoupling limits its accuracy near gliomas ([Strangis et al. 2026](https://doi.org/10.1177/0271678X261496134)), so FC measured
  around a tumour partly reflects its vessels.
- *Generalization.* Evidence on intratumoural function comes mostly from lower-grade, IDH-mutant
  gliomas ([Guarracino et al. 2025](https://doi.org/10.1016/j.nicl.2025.103846)); group-level network findings often do not
  reflect individual patients ([Douw et al. 2026](https://doi.org/10.1038/s41582-025-01171-x)).
- *The study's own boundary.* The article states that structural MRI does not capture the
  information-processing substrate or connectivity, for which diffusion tractography and BOLD fMRI
  are needed, and that its functional-anatomy groups are not tractography or eloquence mapping.

**Neuroscience and clinical track.** A hit-plot is a necessary first step: it says which regions and
which white matter the visible tumour occupies. To reason about cognition it must be combined with
the patient's networks (tractography, carefully interpreted fMRI, stimulation mapping) and with
measured performance.

**Modeling track.** In graph terms, a hit-plot is a node-level perturbation vector, while
disconnection is an edge-level perturbation. Mapping one to the other requires a connectome, either
normative (with the LNM caveats) or individual (with tractography's own false positives and
negatives). Notebook 11 shows how removing edges changes network measures on a phantom.

## 8. Beyond neurons: glia, fields and brain waves

**Why this section.** Connectionism treats the neuron and its synapses as the units of computation.
Recent work, and some bold hypotheses, add glial cells, extracellular fields and rhythmic population
activity. Each has a different level of evidence, and some matter specifically at a glioma border,
where the "glia" are partly tumour.

| Claim | Supporting evidence | Criticism or limits | Status |
| --- | --- | --- | --- |
| Astrocytes take part in information processing and memory | Review of astrocyte roles in synaptic plasticity, network oscillations and memory consolidation ([Santello, Toni & Volterra 2019](https://doi.org/10.1038/s41593-018-0325-8)); noradrenaline's effects on synapses were fully gated by astrocytic adrenergic receptors and calcium, acting through ATP/adenosine ([Lefton et al. 2025](https://doi.org/10.1126/science.adq5480)) | Mechanisms are "not fully understood"; much evidence is from animal models | Established in part; mechanisms under study |
| Astrocyte networks could store memories | A theoretical neuron–astrocyte model forms a high-capacity associative memory, with a Dense Associative Memory and a Transformer as limiting cases ([Kozachkov et al. 2025](https://doi.org/10.1073/pnas.2417788122)) | Purely theoretical; no direct test in tissue | Hypothesis |
| Glioma cells integrate into neural circuits | Neuron–glioma synapses drive tumour growth ([Venkatesh et al. 2019](https://doi.org/10.1038/s41586-019-1563-y); [Venkataramani et al. 2019](https://doi.org/10.1038/s41586-019-1564-x)); tumour functional connectivity relates to survival and language ([Krishna et al. 2023](https://doi.org/10.1038/s41586-023-06036-1)); review in [Monje 2025](https://doi.org/10.1016/j.neuron.2025.07.012) | Mostly mechanistic models and selected surgical cohorts; therapeutic translation in progress | Established mechanism; clinical relevance under study |
| Extracellular fields influence neurons (ephaptic coupling) | A model of the field near prefrontal cortex found field-to-neuron interactions correlating with trial-by-trial changes in oscillatory power ([Pinotsis & Miller 2026](https://doi.org/10.1093/cercor/bhag098)) | Inferred from LFP and a model, not from direct manipulation; notebook 08 shows that a potential is not a field | Evidence growing; size of effect in vivo debated |
| Brain waves carry computation ("analog cognition") | Perspective: bidirectional interactions between spiking and brain waves could support analog computation, flexible routing and consciousness ([Miller, Brincat & Roy 2026](https://doi.org/10.1523/JNEUROSCI.0711-26.2026)) | Opposing views treat oscillations as "exhaust fumes"; [van Bree et al. 2025](https://doi.org/10.1016/j.tics.2024.12.003) separate measurements from processes and conclude that fields are causally relevant while field-potential *signals* can carry information without being causal | Hypothesis with partial support |

**Neuroscience and clinical track.** For a patient, these mechanisms are not yet measurable in
routine care. They matter because they suggest why two patients with similar hit-plots can differ in
cognition: local astrocyte and potassium handling (notebook 07), neuron–glioma coupling (notebooks 05
and 06), and altered field potentials and oscillations near the border (notebook 08) may all change
how a network performs. Each is a reason to measure function directly rather than infer it from
location.

**Modeling track.** Adding glia and fields changes the model class: from weighted graphs of spiking
units to coupled systems with slow glial state variables and spatially continuous fields. The
practical lesson for prediction is restraint: with tens to hundreds of patients, each added mechanism
needs a measurement that constrains it, or it adds only parameters.

## 9. Measuring cognition, and a proposed analysis

**Why measure.** If cognition is distributed across networks of neurons, glia and fields, operating
over milliseconds to months, then no image-derived map can stand in for it. It has to be measured.
Neuropsychological tests sample different abilities that draw on different networks: reaction time,
sustained attention, planning and working memory. CANTAB is a computerized battery that does this
in a standardized way.

**What notebook 10 does.** It analyses a public CANTAB release ([OSF](https://osf.io/5kfw3/)) from a study of
patients with glioma or meningioma before and after surgery, with a caregiver comparison group
([Aerts et al. 2019](https://doi.org/10.1002/pon.5195); MRI resource [Aerts et al. 2022](https://doi.org/10.1038/s41597-022-01806-4)). It reads RTI
(reaction time), RVP (sustained attention), SOC (planning) and SSP (spatial span) at their released
precision and asks when a change between visits exceeds measurement error.

**What must not be done.** That cohort is separate from UCSF-PDGM and LUMIERE. Its MRI has T1,
diffusion and BOLD, whereas the published segmenter needs T1, T1c, T2 and FLAIR; missing contrasts
must not be substituted, and cohorts must not be interchanged. No outcome is validated for the
segmentation study.

**Proposed study (not performed).**
- *Question:* do network features (disconnection indices from a reviewed hit-plot and tractography)
  add held-out information about later cognition beyond tumour volume, location, age and treatment?
- *Design elements from the article's own limitations:* a pre-registered, low-dimensional analysis;
  the article names an expanded leak-safe pool of about 110 UCSF-PDGM baseline subjects and a
  regularized Cox model with forced clinical covariates as a possible survival analysis, which would
  require rerunning the whole chain.
- *For cognition:* repeated testing with reliable-change criteria (notebook 10), reviewed masks and
  anatomy, and an explicitly linked cohort, as outlined in the
  [prediction design](../docs/cognition-prediction-study.md).
- *Measured-function controls:* tissue that looks involved may still be functional; stimulation
  mapping or carefully interpreted fMRI is needed before claiming a functional consequence.
