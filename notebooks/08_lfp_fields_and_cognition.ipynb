**Why this matters.** Neurons and glioma cells are now known to talk to each other electrically.
Neuronal activity drives the growth of gliomas, and gliomas in turn remodel neural circuits to increase
their excitability, which contributes to seizures and cognitive deficits
([Monje 2025](https://doi.org/10.1016/j.neuron.2025.07.012)). In patients, intracranial recordings show that tumour-infiltrated cortex
still takes part in language tasks, but in an altered way ([Aabedi et al. 2021](https://doi.org/10.1073/pnas.2108959118);
[Krishna et al. 2023](https://doi.org/10.1038/s41586-023-06036-1)). Several hypotheses go further and propose that the *electric
fields* generated by brain activity act back on neurons ([Anastassiou et al. 2011](https://doi.org/10.1038/nn.2727)),
organize cognition ([Miller, Brincat & Roy 2026](https://doi.org/10.1523/JNEUROSCI.0711-26.2026)), or even tune the cytoskeleton inside
cells ([Pinotsis, Fridman & Miller 2023](https://doi.org/10.1016/j.pneurobio.2023.102465)). Each of these ideas starts from the same
measurement: a voltage recorded by an electrode. This notebook asks what that voltage can and cannot
tell us, before anything is built on it.

**Why the tumour–brain interface.** The border zone, where infiltrating glioma cells meet neurons,
is where these questions become most pressing. Neurons there form synapses onto glioma cells
([Venkataramani et al. 2019](https://doi.org/10.1038/s41586-019-1564-x); [Venkatesh et al. 2019](https://doi.org/10.1038/s41586-019-1563-y)); slices of
the peritumoral margin from most patients generate spontaneous epileptiform discharges
([Pallud et al. 2014](https://doi.org/10.1126/scitranslmed.3008065)); and the surgeon decides there how much tissue to remove, partly
guided by intraoperative ECoG, which differs between tumour, peritumoral and healthy cortex
([Díaz-Fernández et al. 2025](https://doi.org/10.1093/neuonc/noaf082)). The interface is also an *electrical* boundary: activity is
suppressed over infiltrated cortex ([Zhao et al. 2026](https://doi.org/10.1016/j.jneumeth.2026.110898)), and tumour tissue conducts current
differently from brain ([Latikka & Eskola 2019](https://doi.org/10.1007/s10439-018-02189-7)). Both change the field, and both can
mislead an electrode. Section 6 builds a synthetic border to show how.

| Gradient | The vector of spatial derivatives; the field is $\mathbf{E}=-\nabla\phi$, with three components. |
| Wavelength λ, wavenumber $k=2\pi/\lambda$ | The spatial period of a sinusoidal profile, and the corresponding angular frequency in space (rad/mm). |
| V/m, µV/mm | Field units; 1 µV/mm = 10⁻³ V/m and 1 mV/mm = 1 V/m. |

| Current density $J$ | Current per unit area (A/m²); in a conductor $J=\sigma E$ (Ohm's law). |
| Normal-current condition | At a border between media, the current crossing it is continuous, $\sigma_1E_{1,n}=\sigma_2E_{2,n}$, so the normal field jumps by the conductivity ratio. |
| Product rule | $(A\sin kx)'=A'\sin kx+Ak\cos kx$: a spatially changing amplitude adds a field term at a border. |

| Central difference | Slope estimate $(\phi_{i+1}-\phi_{i-1})/(2\Delta x)$ from two neighbouring samples. |
| Truncation error | Error from replacing a derivative by a finite difference; for a central difference it scales as $\Delta x^2$. |
| Noise amplification | Differentiation divides sample noise by the spacing, so noise grows as $1/\Delta x$. |
| RMS | Root mean square: the square root of the mean squared error. |
| Aliasing, Nyquist limit | With fewer than two samples per wavelength a wave cannot be represented and appears as a different, slower wave or vanishes. |
| Forward and inverse problem | Computing potentials from known sources, and inferring sources from measured potentials; the inverse is not unique without extra assumptions ([Einevoll et al. 2013](https://doi.org/10.1038/nrn3599)). |
| Lead field | The sensitivity of an electrode to sources at each location, computed from a head or tissue model ([Willis et al. 2026](https://doi.org/10.1016/j.neuroimage.2026.121747)). |

| Persistence baseline | Predicting that follow-up equals baseline; the minimum a predictive model must beat. |

## Background: from a voltage trace to cells, tumours and cognition

**Neuroscience and clinical track.**
- *What the LFP is.* Every transmembrane current, synaptic, spiking or other, adds to the potential
  in the surrounding fluid. Synaptic currents are the largest contributor, but sodium and calcium
  spikes, afterpotentials and other currents also shape it ([Buzsáki, Anastassiou & Koch 2012](https://doi.org/10.1038/nrn3241)).
  The LFP contains frequencies up to a few hundred hertz and is harder to interpret than spikes
  because it mixes many sources ([Einevoll et al. 2013](https://doi.org/10.1038/nrn3599)).
- *How local it is.* The geometry of the sources matters more than their strength, much of a
  recorded LFP originates far from the electrode, and its polarity does not tell excitation from
  inhibition ([Herreras 2016](https://doi.org/10.3389/fncir.2016.00101)). The "listening area" grows from multi-unit activity to
  CSD to LFP, and the LFP can spread passively for more than a centimetre
  ([Kajikawa & Schroeder 2011](https://doi.org/10.1016/j.neuron.2011.09.029)).
- *Why fields might matter.* In rat cortical slices, an applied extracellular field changed somatic
  membrane potentials by less than 0.5 mV yet strongly entrained action potentials, especially for
  slow (below 8 Hz) fluctuations ([Anastassiou et al. 2011](https://doi.org/10.1038/nn.2727)). Weak fields in the
  range of endogenous fields entrained the slow oscillation in vitro, so the field generated by a
  network could feed back on it ([Fröhlich & McCormick 2010](https://doi.org/10.1016/j.neuron.2010.06.005)). In humans, intracranial
  measurements during scalp stimulation at 2 mA found cortical fields of up to 0.8 V/m
  ([Huang et al. 2017](https://doi.org/10.7554/eLife.18834)).
- *Why this matters near a glioma.* Gliomas form synapses with neurons, are driven by neuronal
  activity and make circuits more excitable ([Monje 2025](https://doi.org/10.1016/j.neuron.2025.07.012)). Glioma-infiltrated cortex still
  synchronizes during speech but recruits a diffuse network, and its signals carried too little
  information to decode word length ([Aabedi et al. 2021](https://doi.org/10.1073/pnas.2108959118)). Language tasks activate
  tumour-infiltrated cortex beyond normally recruited regions, and functional connectivity between
  tumour and brain is associated with shorter survival and poorer language performance
  ([Krishna et al. 2023](https://doi.org/10.1038/s41586-023-06036-1)). New tools for such studies include intraoperative Neuropixels
  recordings, done in 56 patients, 19 of them with tumours, without complications
  ([Chung et al. 2026](https://doi.org/10.3171/2025.4.JNS241162)), and photograph-guided localization of ECoG electrodes in tumour
  patients ([Oten et al. 2025](https://doi.org/10.1016/j.jneumeth.2025.110588); review: [Herr et al. 2026](https://doi.org/10.1093/noajnl/vdaf109)).
- *The tumour–brain interface as an electrophysiological zone.*
  - *Synapses and tumour networks.* Glioma cells receive AMPA receptor-mediated synaptic input from
    neurons ([Venkatesh et al. 2019](https://doi.org/10.1038/s41586-019-1563-y)), including on tumour microtubes, and this input
    drives invasion and growth ([Venkataramani et al. 2019](https://doi.org/10.1038/s41586-019-1564-x)). The tumour network
    has its own rhythm: a small population of hub cells shows autonomous periodic calcium oscillations
    that depend on the potassium channel KCa3.1 ([Hausmann et al. 2023](https://doi.org/10.1038/s41586-022-05520-4)).
  - *Excitability of the surrounding cortex.* In slices from the peritumoral margin, 69% of patients'
    tissue generated interictal-like discharges near areas of infiltration, sustained by glutamate and
    by depolarizing GABA after changes in chloride cotransporters ([Pallud et al. 2014](https://doi.org/10.1126/scitranslmed.3008065)).
    Pyramidal neurons in glioma-infiltrated human cortex were more excitable in high- than in
    low-grade tumours, and this network activity increased glioma proliferation
    ([McAlpine et al. 2026](https://doi.org/10.1038/s41593-025-02149-0)). In patients, TMS showed increased cortical excitability at
    the border of IDH-mutant gliomas, detectable well beyond the tumour edge
    ([Stark et al. 2026](https://doi.org/10.1093/noajnl/vdag071)), and tumour somatic mutation profiles were associated with EEG
    hyperexcitability ([Tobochnik et al. 2024](https://doi.org/10.1093/neuonc/noad176)).
  - *What electrodes see at the border.* During awake surgery in 29 patients, ECoG contacts on tumour
    showed reduced power in all bands except delta, which increased, while peritumoral cortex had more
    relative beta activity and steeper 20–40 Hz slopes ([Díaz-Fernández et al. 2025](https://doi.org/10.1093/neuonc/noaf082)). Power
    fell with the depth of cortical invasion ([Zhao et al. 2026](https://doi.org/10.1016/j.jneumeth.2026.110898)). Yet infiltrated cortex can
    still take part in cognitive circuits: it modulated high-gamma power with cognitive effort, and its
    connectivity to the dorsal attention network related to long-term attention outcomes
    ([Mandal et al. 2024](https://doi.org/10.1016/j.cortex.2024.01.004)).
  - *The border is also a conductivity boundary.* Intraoperative resistivity averaged 160 Ω·cm in
    low-grade and 498 Ω·cm in high-grade gliomas (about 0.63 and 0.20 S/m), differing by 40–330% from
    surrounding tissue ([Latikka & Eskola 2019](https://doi.org/10.1007/s10439-018-02189-7)), and MRI-based conductivity was higher in
    gliomas than in normal parenchyma ([Tha et al. 2018](https://doi.org/10.1007/s00330-017-4942-5)).

**Modeling track.** In a volume conductor of conductivity σ, the potential produced by transmembrane
currents obeys a Poisson equation, $\nabla\cdot(\sigma\nabla\phi)=-C$, where $C$ is the current-source
density. Three consequences frame this notebook.
- *Fields are derivatives.* The field is $\mathbf{E}=-\nabla\phi$; each component needs spatial
  sampling along its direction.
- *Sources are second derivatives.* In a homogeneous medium $C=-\sigma\nabla^2\phi$, which is why
  CSD analysis differentiates twice and is even more sensitive to spacing, source size and
  conductivity jumps ([Pettersen et al. 2006](https://doi.org/10.1016/j.jneumeth.2005.12.005)).
- *Forward models are practical.* Biophysical forward modelling is how LFPs are interpreted
  ([Einevoll et al. 2013](https://doi.org/10.1038/nrn3599)). Recent work computes LFP and EEG from population firing rates
  with kernels, with an error that depends on how heterogeneous the kernels are and on spike
  correlations ([Ness et al. 2025](https://doi.org/10.1371/journal.pcbi.1012303)), and optimizes electrode placement for information
  capacity using subject-specific finite-element lead fields ([Willis et al. 2026](https://doi.org/10.1016/j.neuroimage.2026.121747)).

This notebook stays with the simplest case, a prescribed potential along a line, because there every
estimate can be checked against an exact answer.

## 1. What an extracellular electrode records

- **Local field potential (LFP).** An electrode in tissue measures a voltage relative to a reference
  electrode. Its low-frequency part, the LFP, reflects the summed transmembrane currents of many
  nearby cells, dominated by synchronous synaptic currents; Buzsáki, Anastassiou & Koch (2012)
  review its origins.
- **Volume conduction.** Currents spread through the conductive extracellular medium, so a
  potential recorded at one point also reflects sources some distance away.
- **The quantity a cell "feels".** A neighbouring cell experiences the extracellular **electric
  field**, the spatial rate of change of the potential. Anastassiou et al. (2011) applied such fields
  to cortical slices and observed effects on membrane potential and spike timing (ephaptic coupling).

So two electrodes with identical voltage traces can sit in very different fields. The rest of the
notebook makes this concrete.

**Neuroscience and clinical track.** Picture a pyramidal neuron in a voltage landscape. If the
potential is the same at both ends of the cell, no current is driven along it, however large that
potential is. What matters is the *difference* in potential across the cell, and the field is that
difference per unit length. A clinical ECoG or sEEG contact therefore reports something about the
activity of the surrounding tissue, but not directly the push on any particular cell. Orientation
matters as well: in awake mice, the apparently heterogeneous effects of cerebellar tDCS on Purkinje
cells were explained by the orientation of each cell's soma and dendrites relative to the field
([Sánchez-León et al. 2025](https://doi.org/10.7554/eLife.100941)). Near a glioma, the cells around an electrode are a mixture
of neurons, glia and infiltrating tumour cells with different shapes and orientations, so the same
field can push them differently.

**Modeling track.** A single electrode samples one value of a scalar function $\phi(\mathbf{r},t)$
relative to $\phi$ at the reference. The field $\mathbf{E}=-\nabla\phi$ is a vector with three
components, each a derivative along one direction. One electrode supplies no derivative at all; a
line of electrodes supplies one component; a three-dimensional array is needed for the full vector.

## 2. Potential and field are different observables
Extracellular potential $\phi$ is recorded relative to a reference, commonly in µV. One field
component is a **spatial** derivative,

$$E_x=-\frac{\partial\phi}{\partial x}\quad\text{in V/m}.$$

A temporal derivative at one electrode cannot replace this gradient.

For the prescribed profile $\phi(x)=A\sin(2\pi x/\lambda)$,

$$E_x(x)=-A\,\frac{2\pi}{\lambda}\cos\!\Big(\frac{2\pi x}{\lambda}\Big),\qquad
|E_x|_{\max}=\frac{2\pi A}{\lambda}.$$

**Units.** 1 µV/mm = 10⁻⁶ V / 10⁻³ m = 10⁻³ V/m. With A = 100 µV, a 10 mm wavelength gives a peak field
of 2π·100/10 = 62.8 µV/mm = 0.063 V/m; a 2.5 mm wavelength gives four times more, 0.251 V/m.

The constructed sampling line spans −5 to 5 mm. These are static spatial illustrations without a
neuronal source-current model, anatomical coordinate transform or brain-wave frequency.

**Neuroscience and clinical track.** Two numbers set the field: the amplitude of the potential and how
quickly it changes across tissue. Activity that is spatially smooth (long wavelength) produces a weak
field even with a large voltage; activity that changes over short distances produces a strong one.
To put 0.251 V/m in cellular terms: across a 20 µm soma it corresponds to a potential difference of
about 5 µV, and along a 1 mm pyramidal cell to about 0.25 mV. These are extracellular differences, not
membrane polarizations, but they give the scale. For comparison, cortical fields during 2 mA scalp
stimulation reached 0.8 V/m in humans ([Huang et al. 2017](https://doi.org/10.7554/eLife.18834)), and in rats at least 1 V/m
(1 mV/mm) was needed to affect neuronal spiking and subthreshold currents
([Vöröslakos et al. 2018](https://doi.org/10.1038/s41467-018-02928-3)). The
synthetic fields here are therefore in the weak range.

**Modeling track.** Write $k=2\pi/\lambda$. Differentiation multiplies each Fourier component by $ik$,
so the field spectrum is the potential spectrum weighted by $k$: differentiation is a high-pass
filter. The potential and its field are in spatial quadrature: where $\phi$ is extremal, $E_x=0$,
and where $\phi$ crosses zero, $|E_x|$ is maximal. The minus sign makes the field point from high to
low potential. Keeping units explicit avoids a classic factor-of-1000 error: µV divided by mm is
mV/m, not V/m.

### Before the simulation: what the next cell does and why

**Why simulate.** In a real recording the true field is unknown, so one cannot tell whether a field
estimate is right. In a simulation we choose the potential, so the exact field is known and every
estimate can be checked. The construction is deliberately minimal: two potentials with *identical*
amplitude (100 µV) that differ only in how quickly they change in space (wavelength 10 or 2.5 mm). If
the field depended only on amplitude, their fields would be equal.

**What the cell does.**
1. It lays out 1001 positions along a 10 mm line, from −5 to 5 mm (spacing 0.01 mm).
2. It evaluates $\phi(x)=A\sin(2\pi x/\lambda)$ for both wavelengths.
3. It estimates $E_x$ from the samples with central differences (`field_from_potential`), converting
   µV/mm to V/m.
4. It checks the peak field against the hand formula $2\pi A/\lambda$ and reports the largest
   difference between the numerical and analytic field.

**Neuroscience and clinical track.** Think of the 1001 positions as an idealized linear probe with
contacts every 10 µm and no noise; sections 4–5 then remove these idealizations.

**Modeling track.** With $\Delta x=0.01$ mm and $k\Delta x\le 0.025$, the leading truncation error of
the central difference is a fraction $(k\Delta x)^2/6\approx10^{-4}$ of the peak, so the reported
maximum error should be of order $10^{-5}$ V/m.

## 3. A reproducible ambiguity
At the centre both potentials are zero, yet their fields differ by a factor of about four. An
electrode's instantaneous voltage, or its amplitude alone, does not determine the field acting
across a cell. Adding a common voltage offset, as a different reference electrode would, does not
change the spatial gradient. The next cell checks both statements numerically.

**Why this matters.** The ambiguity is not a quirk of sinusoids; it is the general relation between a
function and its derivative. It decides what any recording can support.

**Neuroscience and clinical track.**
1. *A value at a point does not determine the slope there.* Two electrodes reading exactly 0 µV can
   sit in fields of −0.063 and −0.251 V/m. Equally, an electrode at a potential peak reads its largest
   voltage while the field there is zero.
2. *The reference cannot rescue it.* Re-referencing adds or subtracts a constant from all channels.
   That changes every voltage but no difference between channels, so the field is untouched. The
   choice of reference matters for interpreting voltages, but a field estimate is immune to a common
   offset, provided all channels share the reference.
3. *Distant sources can masquerade as local fields.* Because LFPs spread far
   ([Kajikawa & Schroeder 2011](https://doi.org/10.1016/j.neuron.2011.09.029)) and much of a recorded LFP is generated remotely
   ([Herreras 2016](https://doi.org/10.3389/fncir.2016.00101)), a smooth gradient across an array can come from a source outside
   it. A spatially linear contamination of 10 µV/mm would add a constant −0.01 V/m to every field
   estimate; unlike a common offset, it is not removed by differencing.
4. *What it means for the cited effects.* The ephaptic effects of Anastassiou et al. (2011) and the
   field thresholds of Huang et al. (2017) and Vöröslakos et al. (2018) are stated in terms of fields
   or membrane changes, not of LFP amplitude. Relating a clinical recording to them requires a field
   estimate, which requires multiple, well-localized contacts.

**Modeling track.**
1. *Non-uniqueness.* For any value $\phi_0$ and any slope $-E_0$ at $x_0$, the function
   $\phi(x)=\phi_0-E_0(x-x_0)$ matches both. A point value constrains the zeroth Taylor coefficient
   only; the field is the first.
2. *Gauge freedom.* $\phi\mapsto\phi+c$ leaves $\mathbf{E}$ unchanged: the potential is defined only up
   to a constant, which is what the reference fixes. The test in the next cell adds 27 µV and checks
   that the field changes by less than 2e-14 V/m.
3. *Time cannot replace space, except under a model.* A temporal derivative $\partial\phi/\partial t$
   at one electrode is a different quantity. For a wave travelling at known speed $v$,
   $\phi(x,t)=f(x-vt)$, it follows that $E_x=\tfrac1v\,\partial\phi/\partial t$, so one electrode would
   suffice. But $v$ and the wave's direction must then be known, and estimating them needs several
   electrodes; human recordings show planar, spiral and concentric traveling waves
   ([Das et al. 2026](https://doi.org/10.1038/s41467-026-71386-z)).
4. *An ill-posed inverse.* Inferring sources or fields from a few potentials is underdetermined
   without a forward model ([Einevoll et al. 2013](https://doi.org/10.1038/nrn3599)). The general lesson echoes the
   identifiability argument of notebook 07: what can be estimated is set by the measurement, not by
   the model.

## 5. Why differentiation amplifies noise

A central difference estimates the slope from two neighbouring samples, $(\phi_{i+1}-\phi_{i-1})/(2\Delta x)$.
Its error has two parts that pull in opposite directions:
- **Truncation:** the straight line between samples misses the curvature. For a sinusoid the RMS error
  is about $A k^3\Delta x^2/(6\sqrt2)$, so it shrinks as $\Delta x^2$.
- **Noise:** independent noise of SD σ on the two samples gives a slope error of SD $\sigma/(\sqrt2\,\Delta x)$,
  which *grows* as samples get closer.

The total error is therefore smallest at an intermediate spacing. The cell below measures the error
on seeded noisy samples (black) and overlays the two terms.

**Neuroscience and clinical track.** A field estimate subtracts two nearby recordings. If the
electrodes are very close, their true potentials are almost equal, and the small true difference is
swamped by the independent noise of each channel. If they are far apart, the straight line between
them misses the shape of the activity. There is a best spacing, and it depends on both the spatial
scale of the activity and the noise. This is a design choice to make before recording, not after.

**Modeling track.** Adding the two independent error terms in quadrature,

$$\mathrm{RMS}^2(\Delta x)=a^2\Delta x^4+\frac{b^2}{\Delta x^2},\qquad a=\frac{Ak^3}{6\sqrt2},\quad
b=\frac{\sigma}{\sqrt2},$$

and setting the derivative to zero gives

$$\Delta x^*=\Big(\frac{b^2}{2a^2}\Big)^{1/6}\propto\Big(\frac{\sigma}{A}\Big)^{1/3}\lambda .$$

The optimum grows only with the cube root of the noise-to-signal ratio and is proportional to the
wavelength. This is the classic bias–variance trade-off of numerical differentiation.

```python
import matplotlib.pyplot as plt

from glioma_ai.neuroglial_interactive import derivative_error_curve, draw_error_curve
from glioma_ai.notebook_display import show_figure

counts = np.array([11, 15, 21, 31, 41, 61, 81, 121, 161, 241, 321, 401])
curves = {
    noise: derivative_error_curve(100.0, 2.5, noise, counts, seed=0, repeats=20)
    for noise in (1.0, 5.0)
}
figure, axes = plt.subplots(1, 2, figsize=(11, 3.8), constrained_layout=True, sharey=True)
for axis, (noise, curve) in zip(axes, curves.items(), strict=True):
    draw_error_curve(axis, curve, noise)
    axis.set_title(
        f"σ = {noise:g} µV: smallest error at Δx ≈ {curve['best_spacing_mm']:.2f} mm", loc="left"
    )
figure.suptitle("Synthetic: A = 100 µV, λ = 2.5 mm; 20 seeded noise realisations per spacing")
show_figure(figure)
for curve in curves.values():
    fine = curve["spacing_mm"] < 0.05
    ratio = curve["empirical_rms_V_per_m"][fine] / curve["noise_rms_V_per_m"][fine]
    assert np.all(np.abs(ratio - 1) < 0.15), ratio
print("At the finest spacings the measured error follows σ/(√2 Δx) within 15%.")
```

spacings the measured error follows $\sigma/(\sqrt2\,\Delta x)$ within 15%.

The general lesson carries over to real recordings even though the numbers do not: a field estimate
needs a design choice of spacing that depends on the wavelength of the activity and the noise, and it
inherits noise from both electrodes.

**Neuroscience and clinical track.** A dense probe is not automatically better for estimating fields.
With contacts closer than the optimum, one can use every second or third contact, or average
neighbouring contacts first. Recent studies treat electrode placement as an optimization problem in
its own right ([Willis et al. 2026](https://doi.org/10.1016/j.neuroimage.2026.121747)).

**Modeling track.** Real noise is often spatially correlated (for example a shared reference or
volume-conducted background). Correlated noise partly cancels in a difference, which shifts the
optimum to smaller spacings; anticorrelated noise does the opposite. The second derivative needed for
CSD amplifies noise as $1/\Delta x^2$ and has a larger optimum (see exercise M5).

## 6. At the tumour–brain interface: a synthetic border

**Why this section.** Sections 1–5 assumed uniform tissue. At a glioma border two things change over
short distances, and both are documented: activity is suppressed over infiltrated cortex
([Díaz-Fernández et al. 2025](https://doi.org/10.1093/neuonc/noaf082); [Zhao et al. 2026](https://doi.org/10.1016/j.jneumeth.2026.110898)), and tissue conductivity
differs between tumour and brain ([Latikka & Eskola 2019](https://doi.org/10.1007/s10439-018-02189-7); [Tha et al. 2018](https://doi.org/10.1007/s00330-017-4942-5)).
The next cell builds the simplest one-dimensional picture of each, with the border at 0 mm, brain on
the left and tumour (shaded) on the right.

**Two constructions.**
- *Suppressed activity (top row of Figure 3).* The 2.5 mm sinusoid of section 2, but with an
  amplitude that falls from 100 µV in brain to 20 µV in tumour over a border of width $w=0.5$ mm:
  $\phi(x)=A(x)\sin(kx)$ with $A(x)=A_b-(A_b-A_t)\tfrac12[1+\tanh(x/w)]$. By the product rule,
  $$E_x=-A(x)\,k\cos(kx)\;-\;A'(x)\sin(kx),$$
  and the second, *border term*, is non-zero only where the amplitude changes.
- *Conductivity step (bottom row).* A steady current of density $J$ crosses the border. Current is
  conserved, so $J=\sigma_bE_b=\sigma_tE_t$ and $E_t=E_b\,\sigma_b/\sigma_t$. With a brain-side field
  of 0.05 V/m and $\sigma_t/\sigma_b=2$, the tumour-side field is 0.025 V/m. The potential is continuous
  but has a kink. The cell samples this potential with electrodes every 0.5 and 2.5 mm and estimates
  the field by central differences, as in section 4.

Both are prescribed constructions: no tumour geometry, oedema shell, three-dimensional current flow or
neural source model, and no values measured in a patient.

**Neuroscience and clinical track.** Think of the top row as an ECoG strip running from healthy cortex
onto infiltrated cortex, and of the bottom row as current crossing the tumour edge, whether it comes
from distant neural sources or from stimulation such as tES or Tumor-Treating Fields.

**Modeling track.** The top row is a test of the product rule on a sampled grid. The bottom row is the
one-dimensional form of the interface conditions of the Poisson equation: $\phi$ is continuous and
the normal current $\sigma\,\partial_n\phi$ is continuous, so the normal derivative jumps by the
conductivity ratio.

```python
from glioma_ai.neuroglial_interactive import draw_tumour_border, tumour_border_fields

border = tumour_border_fields(
    amplitude_brain_uV=100.0,
    amplitude_tumour_uV=20.0,
    wavelength_mm=2.5,
    border_width_mm=0.5,
    field_brain_V_per_m=0.05,
    conductivity_ratio=2.0,
    coarse_spacings_mm=(0.5, 2.5),
)
figure = plt.figure(figsize=(11, 6.5), constrained_layout=True)
draw_tumour_border(figure, border)
figure.suptitle(
    "Synthetic tumour border at 0 mm: brain left, tumour shaded right; prescribed values"
)
show_figure(figure)
x_border = border["positions_mm"]
np.testing.assert_allclose(
    border["activity_numerical_field_V_per_m"], border["activity_field_V_per_m"], atol=1e-4
)
brain_peak = np.abs(border["activity_field_V_per_m"][x_border < -2]).max()
tumour_peak = np.abs(border["activity_field_V_per_m"][x_border > 2]).max()
print(f"Activity: largest field {brain_peak:.3f} V/m in brain, {tumour_peak:.3f} V/m in tumour")
print(f"Largest border term: {np.abs(border['activity_border_term_V_per_m']).max():.3f} V/m")
p = border["parameters"]
print(f"Conductivity step: {p['field_brain_V_per_m']:.3f} → {p['field_tumour_V_per_m']:.3f} V/m")
for spacing, estimate in border["coarse"].items():
    at_border = estimate["estimated_field_V_per_m"][estimate["positions_mm"] == 0][0]
    print(f"Electrodes every {spacing:g} mm: estimate at the border {at_border:.4f} V/m")
```

**Interpretation.**
- *A lower voltage over a tumour has more than one explanation.* In the top row the field is lower
  because activity is lower. But for the same currents, higher conductivity also lowers potentials: in
  a uniform medium the potential of a current source $I$ at distance $r$ is $I/(4\pi\sigma r)$, so
  doubling σ halves every amplitude and quarters the power. Since gliomas conduct differently from
  surrounding tissue (Latikka & Eskola 2019; Tha et al. 2018), part of a reduced ECoG power over tumour
  could in principle reflect conductivity rather than less activity. This is a caution for
  interpretation, not a claim about the cited ECoG findings; separating the two needs conductivity
  measurements or patient-specific forward models.
- *The border creates field structure of its own.* Even without any change in conductivity, a change
  of activity across the border adds the border term.
- *An electrode pair across the border measures a mixture.* Its estimate is the length-weighted mean
  of the fields on the two sides and equals neither.

**Neuroscience and clinical track.**
- *Interpreting border recordings.* A contact's position relative to the tumour edge matters at the
  millimetre scale. After the dura is opened, brain shift degrades the registration of preoperative
  MRI to the patient ([Zhao et al. 2026](https://doi.org/10.1016/j.jneumeth.2026.110898)), and expert ECoG localization in tumour patients
  deviated by 4.3 mm on average ([Oten et al. 2025](https://doi.org/10.1016/j.jneumeth.2025.110588)), more than the ramp width of the 0.5 mm
  spacing here.
- *Stimulation near tumours.* The same physics governs applied fields. In realistic rat glioma models,
  higher oedema conductivity was associated with weaker alternating fields ([Nguyen et al. 2024](https://doi.org/10.1088/1361-6560/ad2e6c)),
  as $E=J/\sigma$ suggests. Removing a standard bone flap above a superficial glioblastoma increased the
  predicted Tumor-Treating Fields strength in the tumour by 60–70% ([Korshoej et al. 2016](https://doi.org/10.1371/journal.pone.0164051)),
  and glioma lesions and craniotomies distort scalp EEG, depending on where the source lies relative to
  the lesion ([Depuydt et al. 2026](https://doi.org/10.1007/s10548-026-01178-7)).
- *Consequence.* Forward models for patients with glioma need to include the tumour, oedema, resection
  cavity and skull defect, and recordings at the border need positions known to within a fraction of
  the spatial scale of interest.

**Modeling track.**
- *Border term.* The amplitude slope peaks at $|A'|_{\max}=(A_b-A_t)/(2w)=80$ µV/mm at $x=0$, but it
  is multiplied by $\sin(kx)$, which is zero at the border itself; the computed peak is 0.039 V/m at
  −0.33 mm. So although $|A'|_{\max}$ scales as $1/w$, the border term grows much less when the border
  sharpens (exercise M8).
- *Mixed estimate.* A central difference with $a$ mm on the brain side and $b$ mm on the tumour side
  returns $(aE_b+bE_t)/(a+b)$. A single pair cannot recover both fields; at least one pair entirely on
  each side, and the border position, are needed.
- *Conductivity and amplitude.* With unknown σ, potentials determine current sources only up to the
  conductivity map: in a uniform medium the source strength $I$ and σ enter only as the ratio
  $I/\sigma$. This is the
  interface version of the identifiability point of section 3.

## 7. Evidence across scales

**Why this section.** Sections 1–6 concerned measurement. Hypotheses about fields, glioma and
cognition chain several steps, and each step has its own kind of evidence. The table goes through the
chain from fields acting on single neurons, through the tumour–brain interface, to cognition and cell
structure.

| Connection | Evidence and interpretation | Required boundary |
| --- | --- | --- |
| Extracellular field → membrane potential/spike timing | Experimental ephaptic effects in rat cortical slices: somatic changes below 0.5 mV, strong entrainment of spikes, [Anastassiou et al. 2011](https://doi.org/10.1038/nn.2727); endogenous-range fields entrain the slow oscillation, [Fröhlich & McCormick 2010](https://doi.org/10.1016/j.neuron.2010.06.005) | Depends on preparation, geometry and field conditions; not proof of cortex-wide cognitive control |
| Applied field strength and orientation → neuronal response | In vivo rats, 0.5–35 V/m: regular-spiking neurons respond at lower fields than fast-spiking ones, and effects grow with field strength, [Chen et al. 2026](https://doi.org/10.1016/j.brs.2026.103053); cell orientation explains heterogeneous tDCS effects, [Sánchez-León et al. 2025](https://doi.org/10.7554/eLife.100941); fields vary across cortical layers in primates, [Lee, Falchier & Opitz 2026](https://doi.org/10.1016/j.brs.2026.103060) | Applied fields; strengths, frequencies and geometries differ from endogenous fields |
| Neurons → glioma cells (synapses, tumour networks) | AMPA receptor-mediated neuron–glioma synapses, partly on tumour microtubes, drive invasion and growth in models, [Venkataramani et al. 2019](https://doi.org/10.1038/s41586-019-1564-x); [Venkatesh et al. 2019](https://doi.org/10.1038/s41586-019-1563-y); rhythmic KCa3.1-dependent hub cells coordinate tumour networks, [Hausmann et al. 2023](https://doi.org/10.1038/s41586-022-05520-4); human cortical network activity increases glioma proliferation, [McAlpine et al. 2026](https://doi.org/10.1038/s41593-025-02149-0) | Chemical and gap-junction signalling, not field effects; mostly models and resected tissue |
| Glioma → excitability of the surrounding cortex | Interictal-like discharges in 69% of patients' peritumoral slices, [Pallud et al. 2014](https://doi.org/10.1126/scitranslmed.3008065); TMS hyperexcitability beyond the border of IDH-mutant gliomas, [Stark et al. 2026](https://doi.org/10.1093/noajnl/vdag071); somatic mutations associated with EEG hyperexcitability, [Tobochnik et al. 2024](https://doi.org/10.1093/neuonc/noad176) | Single-centre or slice studies; excitability is not a field measurement |

| Traveling waves → memory and behaviour | Human intracranial recordings, [Das et al. 2026](https://doi.org/10.1038/s41467-026-71386-z); EEG, [Luo & Ester 2025](https://doi.org/10.1073/pnas.2415573122); direction-specific stimulation improved performance, [Lee et al. 2026](https://doi.org/10.1073/pnas.2527296123) | No glioma-specific studies; the stimulation study shows that imposed wave patterns can matter, not that endogenous ones do |
| Synapses/spikes/waves → organized cognition | Proposed hybrid framework, [Miller, Brincat & Roy 2026](https://doi.org/10.1523/JNEUROSCI.0711-26.2026); model-based evidence for ephaptic coupling in memory, [Pinotsis & Miller 2023](https://doi.org/10.1093/cercor/bhad251) | Theoretical review and model-based inference; no glioma-specific validation |
| Extracellular activity → cytoskeletal organization | Cytoelectric-coupling hypothesis, [Pinotsis, Fridman & Miller 2023](https://doi.org/10.1016/j.pneurobio.2023.102465) | Direct tests need electrophysiology linked to molecular imaging; no assumed LFP-to-microtubule coefficient |
| Connexin-43 ↔ microtubules in glioma stem-like cells | Perturbation experiments, [Smyth et al. 2025](https://doi.org/10.1038/s41419-025-07514-2) | Cytoskeletal relevance does not establish LFP-driven remodeling |
| Applied 200 kHz fields → glioma cell division and survival | Spindle disruption in cell lines, [Kirson et al. 2004](https://doi.org/10.1158/0008-5472.can-04-0083); randomized trial, [Stupp et al. 2017](https://doi.org/10.1001/jama.2017.18718) | Intermediate-frequency applied fields; says nothing about endogenous fields below a few hundred hertz |
| Applied direct-current fields → glioma cell migration | Electrotaxis of glioma cells, with direction depending on cell state (differentiated cells versus stem-like cells), [Clancy et al. 2021](https://doi.org/10.1016/j.yexcr.2021.112736), and on the matrix (2D versus 3D), [Huang et al. 2016](https://doi.org/10.1038/srep21583); PI3K/AKT/mTOR signalling involved, [Lyon et al. 2019](https://doi.org/10.1038/s41598-019-41505-6) | In vitro applied fields; no evidence that endogenous fields at the interface guide invasion |

**Terminology:** intracellular microtubules are tubulin polymers. Tumour microtubes are cell
projections linking glioma networks ([Osswald et al. 2015](https://doi.org/10.1038/nature16071)). An
astrocyte-like malignant-cell program is not a non-malignant astrocyte.

Read the table from top to bottom as a chain. Only the first two rows are experimental effects of
fields on neurons, and the last two of applied fields on glioma cells; the interface rows document
synaptic, excitability and conductivity changes, not field effects; the cytoelectric row is a
hypothesis.

**Neuroscience and clinical track.** The strongest evidence is at the two ends: fields demonstrably
affect neurons in controlled preparations, and gliomas demonstrably change circuits and cognition in
patients. The interface is well characterized synaptically and in terms of excitability, but none of
the studies cited here measured the endogenous fields there. The middle of the chain, from endogenous fields at the
border to tumour-cell behaviour, cell structure or cognitive outcome, is untested. Tumor-treating fields show that applied fields can affect glioma cells
clinically: in a randomized open-label trial of 695 patients, adding them to temozolomide prolonged
median overall survival from 16.0 to 20.9 months ([Stupp et al. 2017](https://doi.org/10.1001/jama.2017.18718)). Their frequency
(200 kHz) and mechanism, forces on charges and dipoles in dividing cells most effective when division
is roughly aligned with the field ([Kirson et al. 2004](https://doi.org/10.1158/0008-5472.can-04-0083)), are far from endogenous LFPs,
so they do not support the cytoelectric hypothesis. They do show that field orientation, frequency
and dose are measurable and matter.

**Modeling track.** Each row is a separate model with its own inputs, outputs and validation. A
chain of models multiplies uncertainties: a weak link anywhere limits the whole chain. Network-level
models of glioma exist: a whole-brain Hopf model fitted to patients' functional connectivity suggested
that tumours increase the influence of interregional interactions
([Alexandersen et al. 2025](https://doi.org/10.1162/netn_a_00426)), and a recent review places symptoms, the connectome
and tumour biology in one multiscale network framework ([Douw, Reijneveld & Mandal 2026](https://doi.org/10.1038/s41582-025-01171-x)).
None of them models extracellular fields explicitly.

## 8. Clinical question and proposed analysis
**Proposed question:** Do neural-dynamics measurements add information about longitudinal language
function beyond reviewed regional tumour burden?

**A border-specific version.** Does activity recorded at the tumour–brain interface (peritumoral
versus tumour versus distant cortex) add information beyond tumour location and burden? This is
plausible because infiltrated cortex can still participate in executive networks, and its
connectivity to the dorsal attention network related to long-term attention outcomes in 17 patients
([Mandal et al. 2024](https://doi.org/10.1016/j.cortex.2024.01.004)), and because ECoG features differ systematically between tumour,
peritumoral and healthy cortex ([Díaz-Fernández et al. 2025](https://doi.org/10.1093/neuonc/noaf082)).

Inputs would include reviewed compartment/parcellation profiles (notebook 09), physiological
recordings with explicit sampling and referencing, and prespecified cognitive assessments. Document
treatment, seizures, antiseizure medication, age, education, subtype and assessment timing. Choose the
endpoint and comparison before fitting; evaluate participants held out from development. EEG/MEG
require their own source/observation models and do not directly measure intracellular microtubule
fields. Sections 4–5 show why electrode spacing and noise belong in such a protocol.

AI could assist code development, source checking and multimodal feature extraction. The imaging
model, assistant and statistical model require separate evaluation. No such analysis, patient
linkage or assistant-benefit experiment is completed here.

**Neuroscience and clinical track.** Why this question is worth asking, and why it is hard:
- *Cognitive outcomes vary widely.* In patients with stable diffuse glioma, four reproducible
  neurocognitive profiles were found (preserved, memory, processing/attention and multi-domain), and
  40% of patients had impaired neurocognitive functioning in both cohorts
  ([Gorter et al. 2025](https://doi.org/10.1093/neuonc/noaf197)).
- *Network measures can track outcome.* After awake resection of language-eloquent glioma, changes in
  functional-network measures were associated with the need for help with instrumental daily
  activities in 15 patients ([Noll et al. 2025](https://doi.org/10.1093/noajnl/vdaf192)). Combining stimulation mapping in 297 patients
  with normative connectivity predicted language impairment and recovery in a separate group better
  than clinical measures ([Coletta et al. 2025](https://doi.org/10.1038/s43856-025-01121-0)).
- *Not every electrophysiological deviation matters cognitively.* In 37 glioma patients measured with
  MEG before surgery and after one year, peritumoral activity became more deviant over time, but
  activity was not associated with executive functioning ([Zimmermann et al. 2025](https://doi.org/10.1093/noajnl/vdaf168)).
  A study must be able to return such a null result.
- *Practical constraints.* Intraoperative recordings are brief and done under surgical conditions;
  electrode positions must be reconstructed, and expert localization of ECoG grids in tumour patients
  deviated by 4.3 mm on average from consensus ([Oten et al. 2025](https://doi.org/10.1016/j.jneumeth.2025.110588)). That is larger than the
  2.5 mm wavelength used here: position errors alone could make a spatial derivative meaningless.
- *Confounds specific to the border.* Seizures and hyperexcitability are common at the interface
  ([Pallud et al. 2014](https://doi.org/10.1126/scitranslmed.3008065); [Stark et al. 2026](https://doi.org/10.1093/noajnl/vdag071)), antiseizure medication changes
  activity, and section 6 shows that a lower amplitude over tumour can reflect conductivity as well
  as activity. Each needs to be recorded or modelled, not ignored.

**Modeling track.** A protocol for adding dynamics features would specify:
- *Observation model.* For each modality, what is measured (scalp potential, magnetic field,
  surface or depth potential), against which reference, with which spatial sampling, and through which
  forward model it relates to sources. For patients with glioma the forward model must include the
  lesion, oedema, cavity and skull defect ([Depuydt et al. 2026](https://doi.org/10.1007/s10548-026-01178-7)).
- *Border coordinates.* Express each contact's position as a signed distance to the reviewed tumour
  border (notebook 09), with its uncertainty, so that border features are defined before outcomes are
  examined.
- *Features before outcomes.* A small number of prespecified features (for example band power,
  aperiodic offset, connectivity or wave direction), computed without access to follow-up scores.
- *Nested comparison.* The same model family with and without the dynamics features, evaluated on
  held-out participants, with uncertainty from resampling participants. The prediction design in the
  empirical anchor below follows this structure.
- *Sensitivity to position.* Propagate electrode-localization uncertainty into every spatial feature;
  sections 4–5 show how spacing errors enter derivatives.

## 9. Testable field hypotheses at the tumour border

A proposed experiment would combine spatially sampled extracellular potentials with molecular imaging
of cytoskeletal dynamics in a defined preparation. Specify measured field amplitudes, gradients,
frequency, orientation and exposure; controls must distinguish direct field effects from changes in
spiking, synaptic/ionic signalling, temperature and viability. Perturbation and concurrent cellular
measurements would be needed to support causality.

**The hypothesis.** Pinotsis, Fridman & Miller (2023) proposed that the electric fields of neural
activity influence molecular-level structures, including the cytoskeleton, so that fields would
"tune" the brain's infrastructure. In glioma, cytoplasmic Cx43 interacts with microtubules and
supports the maintenance and tumorigenicity of stem-like cells ([Smyth et al. 2025](https://doi.org/10.1038/s41419-025-07514-2)), and
glioma cells are coupled into networks by tumour microtubes ([Osswald et al. 2015](https://doi.org/10.1038/nature16071)). A
glioma-specific version of the hypothesis would ask whether the fields generated by neuronal activity
near a tumour change cytoskeletal dynamics in tumour cells. *This is a proposal; no study cited here
tested it.*

**Two neighbouring hypotheses at the border.** The same experimental logic applies to two other
possible field effects on glioma cells, each with a documented starting point.
- *Membrane polarization.* Depolarizing glioma membranes optogenetically promoted proliferation in vivo
  ([Venkatesh et al. 2019](https://doi.org/10.1038/s41586-019-1563-y)). Whether endogenous fields at the interface polarize glioma
  membranes enough to matter is unknown; section 2 gives the scale (a few microvolts across a cell
  body at 0.25 V/m).
- *Directed migration.* Glioma cells migrate directionally in applied direct-current fields, and the
  direction differs between differentiated and stem-like cells ([Clancy et al. 2021](https://doi.org/10.1016/j.yexcr.2021.112736)) and
  between 2D and 3D matrices ([Huang et al. 2016](https://doi.org/10.1038/srep21583)). Whether slow fields at the border
  could bias invasion is untested.

**Application-wise: why it would matter.**
- *Mechanism.* Neuronal activity is known to drive glioma growth through paracrine signalling and
  synapses ([Monje 2025](https://doi.org/10.1016/j.neuron.2025.07.012)). A field-mediated route would be a third, contactless channel,
  acting wherever fields are strong enough, not only where synapses exist. At the border, where
  neurons are hyperexcitable (see the background section), it would add to the synaptic routes.
- *Invasion.* If border fields biased migration, the direction of field gradients relative to the
  tumour edge would become a candidate factor in infiltration, which is what limits resection.
- *Treatment.* Applied fields already affect glioma cells clinically (Tumor-Treating Fields, section 7).
  If endogenous fields at much lower frequencies also acted on tumour cells, field orientation and
  strength near the tumour could become a measurable, and possibly modifiable, factor; section 6 shows
  that tumour conductivity, oedema and skull defects would shape that field.
- *Cognition.* If tumour networks respond to fields, the same activity that supports cognition could
  feed tumour behaviour, linking this question to the circuit–cognition findings of section 8. That
  link would need its own evidence.

**Method-wise: what an informative experiment needs.**
1. *Preparation.* A defined system in which both fields and cells can be observed, for example
   cultured glioma stem-like cells, organotypic slices with implanted tumour cells, or brain
   organoids with invading tumour cells. Each trades realism against control; slices and organoids
   contain a real tumour–brain border, so cells can be analysed by their distance to it, and the
   conductivity of the tumour region should be measured (section 6).
2. *Field delivery and measurement.* Apply fields through electrodes outside the tissue at
   prespecified strengths (V/m), frequencies, durations and orientations, and *measure* the field in
   the tissue with a multi-contact array designed with sections 4–5 in mind (spacing chosen for the
   expected wavelength and noise). Report measured, not nominal, fields.
3. *Cellular readout.* Live imaging of microtubule dynamics or of Cx43 localization, membrane
   potential, or migration tracks, with quantification decided before the experiment, and cell
   orientation relative to the field and to the border recorded for every cell (compare [Sánchez-León et al. 2025](https://doi.org/10.7554/eLife.100941) for neurons and
   [Kirson et al. 2004](https://doi.org/10.1158/0008-5472.can-04-0083) for dividing tumour cells).
4. *Controls that separate direct from indirect effects.* Sham stimulation with identical handling;
   blocking neuronal spiking and synaptic transmission pharmacologically, so that a remaining effect
   cannot be due to changed neuronal activity; temperature and pH monitoring; electrode electrochemistry
   checks; viability assays; and field orientations rotated by 90°, since a direct field effect should
   depend on orientation.
5. *Dose–response and blinding.* Several field strengths spanning endogenous and stimulation ranges
   (compare [Chen et al. 2026](https://doi.org/10.1016/j.brs.2026.103053)), with analysts blinded to condition and a prespecified
   primary outcome.
6. *Perturbation of the proposed mediator.* Interfere with the Cx43–microtubule interaction (as in
   Smyth et al. 2025) and ask whether any field effect disappears.

**Neuroscience and clinical track.** A correlation between field and cytoskeleton in tissue would not
be enough: active neurons produce fields *and* release growth factors and neurotransmitters, so the
field might only be a marker of activity. Only experiments that change the field while holding
neuronal activity fixed, or that block the proposed mediator, can separate the routes.

**Modeling track.** The experiment estimates a dose–response function $R(E,f,\theta)$ of field
amplitude, frequency and orientation. Two design consequences follow. First, the field $E$ is an
estimated input with error (sections 4–5); error in the input attenuates estimated slopes
(errors-in-variables), so the measurement error should be quantified and propagated. Second, an
orientation dependence $R\propto\cos\theta$ is a signature of a direct field effect that indirect,
activity-mediated routes would not share.

## Empirical cognitive anchor

Notebook 10 checks the original CANTAB release and illustrates
paired observations. It supplies behavioral endpoints alongside field hypotheses. The
[prediction design](../docs/cognition-prediction-study.md) compares baseline cognition/anatomy with
added network features in a larger linked cohort. CANTAB/BOLD cannot identify LFP–microtubule or
astrocytic mechanisms.

**Why an anchor.** This notebook works with synthetic potentials and cited hypotheses. Cognitive
claims ultimately have to be tested against measured cognition in patients, with its own units,
precision and measurement error. The anchor is the place in the perspective where that happens.

**Application-wise: what the empirical resource offers.**
- *Source.* A public release of CANTAB scores collected before brain-tumour surgery and at follow-up
  ([OSF release](https://osf.io/5kfw3/)), from the study of [Aerts et al. 2019](https://doi.org/10.1002/pon.5195), with an associated
  multimodal MRI resource ([Aerts et al. 2022](https://doi.org/10.1038/s41597-022-01806-4)). The behavioural and MRI cohorts differ;
  linking them needs an explicit ID crosswalk and imaging QC (notebook 10).
- *Measures.* Five-choice reaction time (RTI, ms), target sensitivity (RVP A′), planning (SOC,
  problems solved in minimum moves) and spatial span (SSP). Each has its own unit and direction of "better" and must never be
  averaged with the others.
- *Change.* Whether a change between visits exceeds test–retest error is judged with the reliable
  change index ([Jacobson & Truax 1991](https://doi.org/10.1037/0022-006X.59.1.12)); notebook 10 shows why compatible reliability
  and reference statistics are needed and are not supplied by the release.
- *What it can and cannot say.* It provides behavioural outcomes before and after surgery. It
  contains no LFP, field, astrocytic or cytoskeletal measurement and cannot identify those
  mechanisms.

**Method-wise: how electrophysiology would enter a cognitive prediction study.** The prediction design
proposes follow-up SSP span as a candidate primary endpoint, with baseline SSP as a predictor, and a
nested sequence of models using only information available before the outcome:

| Model | Adds | Role of this notebook |
| --- | --- | --- |
| Persistence | Baseline SSP only | Reference that every model must beat |
| Clinical baseline | A small frozen set of preoperative clinical variables | — |
| Anatomy | A prespecified tumour burden/location feature (notebook 09) | — |
| Network/dynamics | A few baseline structural or functional connectivity summaries (notebooks 11–12) | — |
| Physiological extension | Appropriately acquired baseline EEG/MEG/ECoG features, in a separate cohort | Sections 1–6 define what "appropriately acquired" means |

For the physiological extension, this notebook implies concrete requirements:
1. *State the observable.* Potential, field estimate or source estimate; they are different features
   (sections 1–3).
2. *Document the sampling.* Electrode positions with their localization error, spacing relative to the
   expected spatial scale of activity, reference, and noise level (sections 4–5;
   [Oten et al. 2025](https://doi.org/10.1016/j.jneumeth.2025.110588)).
3. *Locate contacts relative to the tumour.* Classify each contact as tumour, peritumoral or distant
   using a reviewed border and a signed distance with its uncertainty, and account for the
   conductivity of tumour and oedema when interpreting amplitudes (section 6;
   [Díaz-Fernández et al. 2025](https://doi.org/10.1093/neuonc/noaf082)).
4. *Prespecify few features.* Choose, for example, band power, aperiodic offset, connectivity or
   wave direction before seeing outcomes, and compare against connectivity-only explanations.
5. *Use the same model family.* Ridge regression with and without the physiological features,
   evaluated on held-out participants with participant-level resampling, so that any gain reflects
   information and not flexibility.
6. *Expect and report null results.* MEG activity did not relate to executive function in a glioma
   cohort ([Zimmermann et al. 2025](https://doi.org/10.1093/noajnl/vdaf168)); a well-designed study should be able to show the
   same for its features.

**Neuroscience and clinical track.** The practical aim is to predict and understand the cognitive
course of individual patients, which matters for surgical planning, rehabilitation and counselling.
Electrophysiology would earn a place only if it adds information beyond baseline cognition, clinical
factors and anatomy; this notebook makes clear that the electrode layout determines whether a feature
can carry that information.

**Modeling track.** The design is a nested-model comparison with a fixed model family, which isolates
added information from added flexibility. Leakage is prevented by freezing endpoint, features and
model budget before outcome analysis and by keeping all observations of a person in one split. With
the small glioma sample of the public release, the design is a plan for a larger linked cohort, not a
validated predictor.

## Summary and exercises

**What this notebook shows.**
- *Potential and field are different observables.* The field is the spatial derivative; a single
  electrode supplies none (sections 1–2).
- *Equal voltage amplitudes can hide a fourfold field difference.* At one point, a potential of 0 µV
  coexists with fields of −0.063 and −0.251 V/m (Figure 1, section 3).
- *A common reference offset does not change the field; a spatial gradient of contamination does*
  (section 3).
- *Estimating a field needs spatial sampling whose spacing trades truncation against noise.* With
  σ = 1 and 5 µV the best spacings are about 0.12 and 0.25 mm for a 2.5 mm wavelength (Figure 2).
- *At a tumour border, suppressed activity lowers the field and adds a border term, and a change in
  conductivity makes the field jump.* An electrode pair across the border returns a mixture (0.0375
  V/m between 0.050 and 0.025 V/m), and a lower amplitude over tumour can reflect conductivity as
  well as activity (Figure 3, section 6).
- *The chain from field to cognition has experimental ends and a hypothetical middle* (section 7).

**What it does not show.**
- Any field value in a patient or near a tumour, or any tumour's real conductivity or geometry.
- Any effect of a field on neurons, glioma cells, microtubules, migration, tumour growth or cognition.
- Which electrophysiological features, if any, predict cognitive outcome.

**Exercises: neuroscience and clinical track.**

1. *Same voltage, different push.*
   - *Why:* the core message of the notebook, and the most common misreading of LFP amplitude.
   - *Task:* in Figure 1, find a position where both potentials read 0 µV and one where the teal
     potential is at its peak. What are the two fields at each? Explain to a clinical colleague why a
     large ECoG voltage does not imply a strong field.
   - *Hints:* at the centre both potentials are 0 µV and the fields are −0.063 and −0.251 V/m. At
     2.5 mm the teal potential is +100 µV and its field is 0, while the orange potential is 0 µV with
     a field of −0.251 V/m. The field depends on how quickly the potential changes, not on its value.

2. *How big is that field for a cell?*
   - *Why:* field values become meaningful only against cell size and against measured thresholds.
   - *Task:* convert 0.063 and 0.251 V/m into the potential difference across a 20 µm soma and along a
     1 mm pyramidal cell. Compare with the somatic changes reported by Anastassiou et al. (2011), the
     0.8 V/m of Huang et al. (2017) and the 1 V/m of Vöröslakos et al. (2018).
   - *Hints:* multiply field by length: 0.251 V/m × 20 µm ≈ 5 µV and × 1 mm ≈ 0.25 mV (1.3 µV and
     0.063 mV for the weaker field). These are extracellular differences, not membrane polarizations;
     the actual polarization depends on morphology and orientation (Sánchez-León et al. 2025). Both
     synthetic fields are below 0.8 V/m.

3. *Designing an array.*
   - *Why:* sections 4–5 turn into a concrete decision whenever a probe or grid is chosen.
   - *Task:* activity of interest has a spatial wavelength of about 2.5 mm and the recording noise is
     about 5 µV. Which contact spacing would you choose for a field estimate? What would you do with a
     probe whose contacts are much closer than that?
   - *Hints:* Figure 2 gives about 0.25 mm. With much closer contacts, use every m-th contact so that
     the effective spacing approaches the optimum, or average neighbouring contacts before
     differencing. Also ask whether the noise is independent between contacts (section 5, modeling
     track).

4. *Reading a glioma ECoG study.*
   - *Why:* published patient recordings are the closest existing evidence, and it matters what they
     do and do not measure.
   - *Task:* for Aabedi et al. (2021) and Krishna et al. (2023), state what was recorded and analysed.
     Could either result be restated as a field near the tumour? What would be needed?
   - *Hints:* both used intracranial potentials, analysed for synchrony, entropy, decoding or
     functional connectivity, not for field estimates. A field would need contact spacing suited to the
     spatial scale of activity and accurate positions: the 4.3 mm mean localization deviation reported
     by Oten et al. (2025) is larger than the 2.5 mm wavelength used here.

5. *A null result.*
   - *Why:* designing studies that can show "no added value" protects against overinterpretation.
   - *Task:* Zimmermann et al. (2025) found no association between MEG activity and executive
     functioning in glioma patients. List possible reasons, and say how the nested design of the
     empirical anchor would treat a similar result for LFP features.
   - *Hints:* consider the observable (broadband power and offset averaged over regions), the outcome
     (two executive tests), the sample (37 patients), and the possibility that network properties
     matter more than local activity, as the authors suggest. In the nested design, the physiological
     model would simply fail to beat the network model on held-out participants, and that would be
     reported.

6. *Tumor-treating fields and the cytoelectric hypothesis.*
   - *Why:* an approved field therapy for glioblastoma is the obvious counterargument, and it should be
     weighed correctly.
   - *Task:* does the survival benefit of Tumor-Treating Fields (Stupp et al. 2017) support the idea
     that endogenous neural fields affect glioma cells? Compare frequency, mechanism and field source.
   - *Hints:* 200 kHz applied fields versus endogenous activity below a few hundred hertz; a proposed
     mechanism of forces on charges and dipoles in dividing cells (Kirson et al. 2004) versus a proposed
     tuning of cytoskeleton; external transducer arrays versus fields generated by neurons. What they do
     share is that orientation and dose are measurable and matter.

7. *Less power over the tumour: less activity or different tissue?*
   - *Why:* reduced ECoG power over infiltrated cortex is a reported finding (Díaz-Fernández et al.
     2025; Zhao et al. 2026), and section 6 shows that conductivity can also lower amplitudes.
   - *Task:* suppose a contact over tumour shows a quarter of the power of a contact over healthy cortex.
     Give two explanations based on section 6, and say which observations or measurements could
     separate them.
   - *Hints:* a quarter of the power is half the amplitude. Explanation one: activity is halved (top
     row of Figure 3, where a fivefold amplitude drop gives a fivefold field drop). Explanation two: in
     a uniform-medium approximation, doubling conductivity halves the potential of the same currents.
     To separate them, measure conductivity (intraoperatively, Latikka & Eskola 2019; with MRI, Tha et
     al. 2018) or build a patient-specific forward model. The spectrum also helps: if conductivity
     does not depend on frequency in the relevant range, it scales all bands by the same factor, so it
     cannot explain the increase in delta together with decreases in other bands reported by
     Díaz-Fernández et al. (2025).

8. *Planning a strip across the border.*
   - *Why:* border recordings are only as informative as their geometry, and surgery limits how many
     contacts and how much time are available.
   - *Task:* using Figure 3, plan a contact layout that could estimate the field on both sides of a
     tumour border and report the border region separately. What information do you need about each
     contact's position, and where would it come from?

**Exercises: modeling track.**

1. *The exact central-difference factor.*
   - *Why:* it explains the undersampling task of section 4 exactly, without simulation.
   - *Task:* show that for $\phi=A\sin(kx)$ the central difference equals the true derivative times
     $\sin(k\Delta x)/(k\Delta x)$. Evaluate it for 21 samples on the 10 mm line with λ = 2.5 mm and
     compare with the leading-order truncation formula.
   - *Hints:* use $\sin(a+b)-\sin(a-b)=2\cos a\sin b$. With Δx = 0.5 mm, $k\Delta x=1.257$ and the
     factor is 0.757, so the estimated peak is 0.190 V/m (explorer: 0.190 V/m). The leading-order
     approximation $1-(k\Delta x)^2/6$ gives 0.737, too pessimistic at this coarse spacing.

2. *Exactly two samples per wavelength.*
   - *Why:* the Nyquist limit is often quoted as "two samples per wavelength", which is not sufficient
     for a derivative.
   - *Task:* explain why 9 samples with λ = 2.5 mm give an estimated field of zero everywhere, and show
     that this does not depend on where the samples fall.
   - *Hints:* Δx = 1.25 mm, so $k\Delta x=\pi$ and $\sin(k\Delta x)=0$. Then $\phi_{i+1}=\phi_{i-1}$
     for every phase, because $\sin(a+\pi)=\sin(a-\pi)$. Shifting the samples by 0.3 mm leaves every
     central difference zero to rounding (about 1e-13 µV).

3. *The optimal spacing.*
   - *Why:* Figure 2 shows an optimum; deriving it shows how it scales.
   - *Task:* derive $\Delta x^*$ from section 5, evaluate it for σ = 1 and 5 µV, and predict how it
     changes if the wavelength doubles or the amplitude halves.
   - *Hints:* $\Delta x^*=(b^2/2a^2)^{1/6}$ gives 0.139 and 0.237 mm, with minimum errors of 0.0062 and
     0.018 V/m (measured: 0.0062 and 0.017 V/m). Since $\Delta x^*\propto(\sigma/A)^{1/3}\lambda$,
     doubling λ doubles it; halving A multiplies it by $2^{1/3}\approx1.26$.

4. *Space from time with a traveling wave.*
   - *Why:* it shows when a single electrode can, in principle, give a field, and what that costs.
   - *Task:* for $\phi(x,t)=A\sin(k(x-vt))$ show that $E_x=\tfrac1v\,\partial\phi/\partial t$. With
     A = 100 µV, λ = 2.5 mm and a frequency of 10 Hz, compute $v$ and the peak field from the peak
     temporal derivative. What would a 20% error in $v$ do?
   - *Hints:* $v=f\lambda=25$ mm/s. The peak of $\partial\phi/\partial t$ is $2\pi fA=6283$ µV/s, and
     dividing by $v$ gives 251 µV/mm = 0.251 V/m, as in Figure 1. A 20% error in $v$ gives a 20% error in
     the field, and the direction of travel must be known; both need several electrodes (Das et al.
     2026).

5. *Current-source density.*
   - *Why:* CSD analysis differentiates twice ([Pettersen et al. 2006](https://doi.org/10.1016/j.jneumeth.2005.12.005)), which makes the
     spacing problem sharper.
   - *Task:* for the second difference $(\phi_{i+1}-2\phi_i+\phi_{i-1})/\Delta x^2$, derive the noise
     SD and the leading truncation error, and find the optimal spacing for σ = 1 and 5 µV with
     A = 100 µV and λ = 2.5 mm.
   - *Hints:* noise SD $\sqrt6\,\sigma/\Delta x^2$; truncation RMS $Ak^4\Delta x^2/(12\sqrt2)$. Minimizing
     the quadrature sum gives $\Delta x^*=(b/a)^{1/4}$ with $a=Ak^4/(12\sqrt2)$ and $b=\sqrt6\,\sigma$:
     0.32 mm and 0.48 mm, about twice the first-derivative optima. At σ = 5 µV the minimum RMS error is
     about 76 µV/mm², 12% of the peak curvature $Ak^2\approx632$ µV/mm². A direct simulation with 200
     noise realisations gave 75 µV/mm² at Δx = 0.5 mm and 196 µV/mm² at 0.25 mm.

6. *Offset versus gradient contamination.*
   - *Why:* reference effects are not always constant across an array.
   - *Task:* add $c+gx$ to every sample, with $c=150$ µV and $g=10$ µV/mm. Which part changes the field
     estimate, and by how much? How could one detect such contamination?
   - *Hints:* the constant changes nothing (a test with `sampled_field` and `field_from_potential`
     gives a difference of about 1e-16 V/m); the gradient adds $-g$, a constant −0.01 V/m bias at every
     sample. It can be detected by recording at sites known to be inactive, by comparing references, or
     with a forward model of distant sources.

7. *An electrode pair across a conductivity border.*
   - *Why:* it shows exactly what a bipolar measurement across a tumour edge returns, and how many
     contacts are needed to undo the mixing.
   - *Task:* derive $E_t=E_b\,\sigma_b/\sigma_t$ from current conservation. For the bottom row of Figure
     3, compute the estimate of a 4 mm pair centred at −1, 0 and +1 mm. Then show that four collinear
     contacts with the border in the middle interval determine $E_b$, $E_t$ and the border position,
     but three do not.
   - *Hints:* a pair with $a$ mm in brain and $b$ mm in tumour returns $(aE_b+bE_t)/(a+b)$: 0.04375,
     0.0375 and 0.03125 V/m. With four contacts $x_1<x_2<x_0<x_3<x_4$, the potential differences give
     $E_b$ from the first pair, $E_t$ from the last, and $x_0$ from the middle one, which equals
     $E_b(x_0-x_2)+E_t(x_3-x_0)$. With three contacts, two unknowns remain in one equation.

8. *How large is the border term?*
   - *Why:* the product rule predicts extra field where activity changes; it is worth checking how
     strongly that depends on the sharpness of the border.
   - *Task:* run `tumour_border_fields` with border widths of 0.25, 0.5 and 1 mm and report the peak of
     `activity_border_term_V_per_m`. Why does it not double each time the width halves?
   - *Hints:* the peaks are 0.043, 0.039 and 0.030 V/m (at −0.18, −0.33 and −0.49 mm), while
     $|A'|_{\max}=(A_b-A_t)/(2w)$ is 160, 80 and 40 µV/mm. The border term is $-A'(x)\sin(kx)$, and here
     the border lies at a zero of $\sin(kx)$: the sharper the border, the more of $A'$ falls where the
     sine is small. A border at a peak of the sine would behave differently; modifying the function to
     shift the phase is a further exercise.
